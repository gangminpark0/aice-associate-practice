## 05 · 숫자형 변수 자세히 보기
교재 **54–55쪽**

```python
# 나이의 평균과 중앙값
mean_age = df['Age'].mean()
print(f"승객 평균 나이: {mean_age:.2f}세")
median_age = df['Age'].median()
print(f"승객 나이 중앙값: {median_age}세")
# 최고령 및 최연소 승객
max_age = df['Age'].max()
min_age = df['Age'].min()
print(f"최고령 승객: {max_age}세, 최연소 승객: {min_age}세")
# 전체 요금의 합계
total_fare = df['Fare'].sum()
print(f"총 탑승 요금: {total_fare:.2f}달러")
# 나이가 기록된 승객 수
age_count = df['Age'].count()
print(f"나이가 기록된 승객 수: {age_count}명")
```

## 09 · IQR로 이상치 찾기
교재 **59쪽**

```python
# 사분위수 계산
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1
# 이상치 경계 설정
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print(f"이상치 최소 경계: {lower_bound:.2f}")
print(f"이상치 최대 경계: {upper_bound:.2f}")
# 경계를 벗어나는 행 선택
outliers = df[(df['Fare'] < lower_bound) | (df['Fare'] > upper_bound)]
print(f"\n총 {len(outliers)}개의 이상치 발견")
print(outliers.head())
```

## 10 · 결측치가 있는 행 삭제
교재 **61쪽**

```python
print("처리 전 데이터 수:", len(df))
print("Age 결측치 수:", df['Age'].isnull().sum())
# Age가 비어 있는 행만 제외
df_dropped = df.dropna(subset=['Age'])
print("처리 후 데이터 수:", len(df_dropped))
```

## 11 · 평균으로 결측치 채우기
교재 **61쪽**

```python
# Age의 결측치를 평균으로 대체
df['Age'] = df['Age'].fillna(df['Age'].mean())
print("Age 결측치 수:", df['Age'].isnull().sum())
```

## 12 · 중앙값과 최빈값으로 채우기
교재 **61쪽**

```python
# 다른 처리 방법과 비교하기 위해 원본 복원
df = df_original.copy()
# 중앙값으로 대체
df['Age'] = df['Age'].fillna(df['Age'].median())
# 최빈값으로 대체
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
print(df[['Age', 'Embarked']].isnull().sum())
```

## 13 · 이상치 행 제외하기
교재 **62쪽**

```python
# IQR 기반 경계 계산
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
# 정상 범위에 해당하는 행만 선택
df_no_outlier = df[(df['Fare'] >= lower) & (df['Fare'] <= upper)]
print(f"원본: {len(df)}행 → 이상치 제거 후: {len(df_no_outlier)}행")
```

## 14 · clip으로 상한 적용
교재 **62쪽**

```python
# Titanic과 별개인 시속 연습 데이터
speed_df = pd.DataFrame({'Speed_Per_Hour': [5, 80, 320, 450]})
# 300보다 큰 값은 300으로 대체
speed_df['Speed_Per_Hour'] = speed_df['Speed_Per_Hour'].clip(upper=300)
print("처리 후 최대값:", speed_df['Speed_Per_Hour'].max())
```

## 15 · 하한·상한과 IQR clip
교재 **63쪽**

```python
# 시속 예제: 하한과 상한을 동시에 설정
speed_df['Speed_Per_Hour'] = speed_df['Speed_Per_Hour'].clip(lower=10, upper=300)
# Titanic 요금의 IQR 경계 계산
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
df['Fare'] = df['Fare'].clip(lower=lower_bound, upper=upper_bound)
print(f"Fare 범위: {df['Fare'].min():.1f} ~ {df['Fare'].max():.1f}")
```
