## 06 · NumPy 배열 만들기
교재 **35쪽**

```python
# NumPy 라이브러리를 'np'라는 별칭으로 불러옵니다.
import numpy as np
# 파이썬 리스트 생성
scores_list = [95, 88, 100]
# 리스트를 NumPy 배열로 변환
scores_array = np.array(scores_list)
print("파이썬 리스트:", scores_list)
print("NumPy 배열:", scores_array)
print("배열의 타입:", type(scores_array))
```

## 07 · 반복문으로 5점 더하기
교재 **36쪽**

```python
# 파이썬 리스트를 사용한 단계별 작업
scores_list = [70, 85, 92, 68]
updated_scores = []  # 결과를 담을 빈 리스트 준비
# scores_list의 모든 점수에 대해 반복 작업을 시작
for score in scores_list:
    # 꺼낸 점수에 5를 더해서, 준비해둔 빈 리스트에 추가
    updated_scores.append(score + 5)
print(updated_scores)
```

## 08 · NumPy 벡터화
교재 **36쪽**

## 09 · 배열 인덱싱
교재 **37쪽**

```python
import numpy as np
scores = np.array([88, 92, 75, 99, 81])
# 첫 번째(0번 인덱스) 원소 꺼내기
first_score = scores[0]
# 네 번째(3번 인덱스) 원소 꺼내기
fourth_score = scores[3]
print("전체 점수:", scores)
print("첫 번째 점수:", first_score)
print("네 번째 점수:", fourth_score)
```

## 10 · 배열 슬라이싱
교재 **38쪽**

```python
import numpy as np
scores = np.array([88, 92, 75, 99, 81])
# 1번 인덱스부터 4번 인덱스 "앞까지" (즉, 1~3번) 잘라내기
middle_scores = scores[1:4]
# 처음부터 3번 인덱스 "앞까지" (즉, 0~2번) 잘라내기
# 시작 인덱스는 생략할 수 있습니다.
first_half = scores[:3]
# 2번 인덱스부터 "끝까지" 잘라내기
# 끝 인덱스는 생략할 수 있습니다.
second_half = scores[2:]
print("1번부터 3번까지:", middle_scores)
print("처음부터 2번까지:", first_half)
print("2번부터 끝까지:", second_half)
```

## 11 · Pandas Series
교재 **39쪽**

```python
import pandas as pd
import numpy as np
# NumPy 배열 (숫자 주소만 있음)
scores_np = np.array([95, 88, 100])
# Pandas Series (인덱스 이름표를 직접 지정)
scores_pd = pd.Series([95, 88, 100], index=['Alice', 'Bob', 'Charlie'])
print("--- NumPy Array ---")
print(scores_np)
print("\n--- Pandas Series ---")
print(scores_pd)
```

## 12 · DataFrame 성적표 만들기
교재 **40–41쪽**

```python
import pandas as pd
# 1. 성적표에 넣을 '열(Column)' 단위로 데이터를 준비합니다.
student_data = {
    '국어': [95, 100, 85],
    '영어': [85, 95, 90],
    '수학': [100, 80, 95]
}
# 2. pd.DataFrame()이라는 틀에 준비된 데이터를 넣어줍니다.
df = pd.DataFrame(student_data, index=['Alice', 'Bob', 'Charlie'])
# 3. 완성된 성적표(DataFrame)를 출력합니다.
print(df)
```

## 13 · 한 개의 열 선택
교재 **41쪽**

```python
# '수학' 점수만 선택
math_scores = df['수학']
print("\n--- 수학 점수 ---")
print(math_scores)
```

## 14 · 여러 열 선택
교재 **42쪽**

```python
# '국어'와 '영어' 점수를 선택
kor_eng_scores = df[['국어', '영어']]
print("\n--- 국어, 영어 점수 ---")
print(kor_eng_scores)
```

## 15 · loc: 이름으로 행과 값 선택
교재 **42쪽**

```python
# 'Bob' 학생의 모든 과목 점수 (행) 선택
bob_scores = df.loc['Bob']
# 'Charlie' 학생의 '영어' 점수만 선택 (행, 열 동시 지정)
charlie_eng_score = df.loc['Charlie', '영어']
print("\n--- Bob의 모든 점수 ---")
print(bob_scores)
print("\nCharlie의 영어 점수:", charlie_eng_score)
```

## 16 · iloc: 위치로 행과 값 선택
교재 **43쪽**

```python
# 1번 위치의 행 (Bob) 선택 (Alice가 0번이므로 Bob은 1번)
second_student = df.iloc[1]
# 0번 위치의 행(Alice)의 2번 위치의 열(수학) 선택
alice_math_score = df.iloc[0, 2]
print("\n--- 1번 위치 학생의 점수 ---")
print(second_student)
print("\n[0, 2] 위치의 값:", alice_math_score)
```
