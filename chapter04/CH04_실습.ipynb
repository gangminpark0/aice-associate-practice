## 01 · 라이브러리와 데이터 준비
교재 **72쪽**

```python
# 데이터 분석 및 시각화에 필요한 기본 라이브러리를 불러옵니다.
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# Titanic 데이터를 원본 파일에서 다시 불러옵니다.
df = pd.read_csv("Part2 Titanic.csv")
print(df.shape)
```

## 02 · 생존자와 사망자 수
교재 **73쪽**

```python
plt.figure(figsize=(8, 6))
# Survived 변수의 각 값(0, 1)이 몇 개인지 세어 막대그래프로 그립니다.
sns.countplot(x='Survived', data=df)
plt.title('생존자 vs 사망자 수 (0: 사망, 1: 생존)')
plt.show()
```

## 03 · 객실 등급별 승객 수
교재 **74쪽**

```python
plt.figure(figsize=(8, 6))
# Pclass 변수의 각 값(1, 2, 3)의 개수를 세어 막대그래프로 그립니다.
sns.countplot(x='Pclass', data=df)
plt.title('객실 등급별 승객 수')
plt.show()
```

## 04 · 승객 나이 분포
교재 **75쪽**

```python
plt.figure(figsize=(10, 6))
# Age 변수에 대한 히스토그램을 그립니다.
sns.histplot(data=df, x='Age', bins=30, kde=True)
plt.title('승객 나이 분포')
plt.xlabel('나이 (Age)')
plt.ylabel('빈도 (Frequency)')
plt.show()
```

## 05 · 객실 등급별 생존율
교재 **76쪽**

```python
plt.figure(figsize=(8, 6))
# Pclass에 따른 Survived의 평균을 막대그래프로 그립니다.
sns.barplot(x='Pclass', y='Survived', data=df)
plt.title('객실 등급에 따른 생존율')
plt.ylabel('생존율 (Survival Rate)')
plt.show()
```

## 06 · 성별 생존율
교재 **77쪽**

```python
plt.figure(figsize=(8, 6))
# Sex에 따른 Survived의 평균을 막대그래프로 그립니다.
sns.barplot(x='Sex', y='Survived', data=df)
plt.title('성별에 따른 생존율')
plt.ylabel('생존율 (Survival Rate)')
plt.show()
```

## 07 · 객실 등급별 요금 분포
교재 **77쪽**

```python
plt.figure(figsize=(10, 8))
# Pclass에 따른 Fare의 분포를 상자 그림으로 그립니다.
sns.boxplot(x='Pclass', y='Fare', data=df)
plt.title('객실 등급별 요금 분포')
plt.ylabel('요금 (Fare)')
plt.show()
```

## 08 · 나이와 요금의 산점도
교재 **78쪽**

```python
plt.figure(figsize=(10, 8))
# Age와 Fare의 관계를 산점도로 그립니다.
sns.scatterplot(x='Age', y='Fare', data=df)
plt.title('나이와 요금의 관계 (산점도)')
plt.xlabel('나이 (Age)')
plt.ylabel('요금 (Fare)')
plt.show()
```

## 09 · 관계와 분포를 함께 보기
교재 **79쪽**

```python
# 나이와 요금의 관계 및 분포를 jointplot으로 그립니다.
sns.jointplot(x='Age', y='Fare', data=df)
plt.show()
```

## 10 · 상관관계 히트맵
교재 **80쪽**

```python
# 상관계수 행렬을 계산합니다.
corr_matrix = df.corr(numeric_only=True)
plt.figure(figsize=(12, 10))
# 히트맵을 생성합니다.
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('변수 간 상관관계 히트맵')
plt.show()
```

## 11 · 나이에 따른 평균 요금
교재 **81쪽**

```python
plt.figure(figsize=(10, 6))
# 나이에 따른 평균 요금의 추세를 선 그래프로 그립니다.
sns.lineplot(x='Age', y='Fare', data=df)
plt.title('나이에 따른 평균 요금 추세')
plt.xlabel('나이 (Age)')
plt.ylabel('평균 요금 (Average Fare)')
plt.show()
```

## 12 · 객실 등급과 성별을 함께 비교
교재 **82쪽**

```python
plt.figure(figsize=(10, 8))
# 객실 등급별 생존율을 성별 색상으로 나눕니다.
sns.barplot(x='Pclass', y='Survived', hue='Sex', data=df)
plt.title('객실 등급과 성별에 따른 생존율')
plt.ylabel('생존율 (Survival Rate)')
plt.show()
```
